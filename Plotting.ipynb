```python
import numpy as np
import scipy as sp
from cellCellAdhesion import explicitNeuralCrest1D

import matplotlib.pyplot as plt
```

```python
L = 200

Meshpoints = 1000

data = np.loadtxt("Y_mu_fl1_mu_ff1_xi_fl1_xi_ff1.txt", delimiter=",")
time_data = np.loadtxt("T_mu_fl1_mu_ff1_xi_fl1_xi_ff1.txt", delimiter=",")

leader_data = data[:1000, :]
follower_data = data[1000:, :]

for time_index, time in enumerate(time_data):
    leader_snapshot = leader_data[:, time_index]
    follower_snapshot = follower_data[:, time_index]

    plt.plot(np.linspace(0, L, Meshpoints), leader_snapshot, label="Leader")
    plt.plot(np.linspace(0, L, Meshpoints), follower_snapshot, label="Follower")
    plt.xlabel("Position")
    plt.ylabel("Density")
    plt.title(f"Time = {time}")
    plt.legend()
    plt.show()
```
